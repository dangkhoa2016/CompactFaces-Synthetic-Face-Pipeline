## 02 · Tìm input và hiển thị chế độ

> **Tiếng Việt**


Ưu tiên `DATASET_DIR`, tiếp theo `COMPACTFACES_DATASET_DIR`, cuối cùng tìm trong Kaggle input. Không lặng lẽ chấp nhận input thiếu file hay nhiều thư mục không rõ. Notebook có thể tự chạy DEMO dù chưa attach dữ liệu.

**Dữ liệu Kaggle:** Version 1 cung cấp trực tiếp thư mục PNG và metadata; FULL đối chiếu SHA256 của từng ảnh thật.

Chỉ chạy FULL sau khi attach Dataset Version 1. Notebook tự dò các thư mục dữ liệu. Muốn xem các file, mở phần **Input** trên Kaggle.


```python
import csv
import hashlib
import json
import time
import zipfile
import re
import io
import struct
import platform
from collections import Counter, defaultdict
from pathlib import PurePosixPath
from datetime import datetime, timezone

SLOTS = ("SOURCE", "P1", "P2", "E1", "E2")
COLLECTIONS = {"CORE": ("core_600png.zip", 600, 120),
               "ADDITIONAL_OPTIONAL": ("additional_optional_335png.zip", 335, 67)}
OPTIONAL_LOCAL_ZIP_FILES = ("metadata.csv", "core_600png.zip",
                   "additional_optional_335png.zip", "collection_builder_187.zip")
def is_extracted_root(base):
    return bool((base / "metadata.csv").is_file()
                and (base / "core_600png/images").is_dir()
                and (base / "additional_optional_335png/images").is_dir()
                and (base / "collection_builder_187/WEB_COLLECTION_BUILDER/previews").is_dir())

REQUIRED_COLS = {"file_name", "identity_key", "collection", "slot", "sha256", "source_sha256",
                 "archive_name", "archive_file_name"}
HEADINGS = {
  "en": {"full":"FULL DATASET", "demo":"DEMO · METADATA ONLY", "pass":"PASS",
         "skipped":"SKIPPED", "partial":"PARTIAL", "bad":"FAIL"},
  "vi": {"full":"DỮ LIỆU ĐẦY ĐỦ", "demo":"DEMO · CHỈ METADATA", "pass":"ĐẠT",
         "skipped":"BỎ QUA", "partial":"CHƯA ĐỦ", "bad":"LỖI"},
}
def say(en, vi):
    print(en if LANGUAGE == "en" else vi)
def require(ok, explanation):
    if not ok:
        raise ValueError(explanation)
def sha256_stream(reader):
    digest = hashlib.sha256()
    while chunk := reader.read(1024 * 1024):
        digest.update(chunk)
    return digest.hexdigest()
def is_sha(value):
    return bool(re.fullmatch(r"[0-9a-f]{64}", str(value)))
def verify_png_header_and_hash(reader, filename, compute_hash):
    """Check each PNG signature and IHDR dimensions, then optionally stream its SHA256."""
    header = reader.read(24)
    require(len(header) == 24 and header[:8] == bytes.fromhex("89504e470d0a1a0a")
            and header[8:12] == bytes.fromhex("0000000d") and header[12:16] == b"IHDR",
            f"Invalid PNG signature/IHDR: {filename}")
    dimensions = struct.unpack(">II", header[16:24])
    require(dimensions == (EXPECTED_PNG_SIZE, EXPECTED_PNG_SIZE),
            f"PNG dimensions mismatch: {filename}: {dimensions}; expected {EXPECTED_PNG_SIZE}x{EXPECTED_PNG_SIZE}")
    if not compute_hash:
        return None
    digest = hashlib.sha256(header)
    for chunk in iter(lambda: reader.read(1024 * 1024), b""):
        digest.update(chunk)
    return digest.hexdigest()

def safe_member(path):
    p = PurePosixPath(str(path))
    return bool(path and not p.is_absolute() and p.parts[0] == "images"
                and ".." not in p.parts and p.suffix.lower() == ".png")

explicit = "" if MODE == "demo" else (DATASET_DIR or os.getenv("COMPACTFACES_DATASET_DIR", "")).strip()
root = Path(explicit).expanduser().resolve() if explicit else None
candidates = []
incomplete_candidates = []
kaggle_inputs = Path(os.getenv("COMPACTFACES_KAGGLE_INPUT_ROOT", "/kaggle/input"))
if not explicit and MODE != "demo" and kaggle_inputs.is_dir():
    # Kaggle can mount dataset versions under a nested folder, and some
    # inputs are symlink-backed. Check each likely path without writing input.
    probes = [kaggle_inputs]
    probes.extend(p.resolve() for p in kaggle_inputs.iterdir() if p.is_dir())
    # Ignore unrelated Kaggle inputs, but reject malformed CompactFaces mounts.
    # A missing Builder directory or metadata must not silently select DEMO.
    inspected_metadata = set()
    for base in probes:
        if not base.is_dir():
            continue
        for metadata in base.rglob("metadata.csv"):
            parent = metadata.parent.resolve()
            if parent in inspected_metadata:
                continue
            inspected_metadata.add(parent)
            if (all((parent / name).is_file() for name in OPTIONAL_LOCAL_ZIP_FILES)
                    or is_extracted_root(parent)):
                candidates.append(parent)
            else:
                with metadata.open(encoding="utf-8-sig", errors="replace") as handle:
                    heading = handle.readline()
                named_for_compactfaces = parent.name.startswith("compactfaces-synthetic-face-dataset")
                has_compactfaces_columns = all(x in heading for x in ("identity_key", "collection", "slot"))
                has_compactfaces_parts = any((parent / part).exists() for part in (
                    "core_600png", "additional_optional_335png", "collection_builder_187",
                    "core_600png.zip", "additional_optional_335png.zip",
                    "collection_builder_187.zip", "source_lineage.jsonl"))
                if named_for_compactfaces or has_compactfaces_columns or has_compactfaces_parts:
                    incomplete_candidates.append(parent)
    candidates = sorted(set(candidates))
    for base in probes:
        # Also catch a named Kaggle mount with missing metadata.csv entirely.
        if (base.is_dir() and base.name.startswith("compactfaces-synthetic-face-dataset")
                and not any(candidate == base or candidate.is_relative_to(base) for candidate in candidates)
                and base not in incomplete_candidates):
            incomplete_candidates.append(base)
    incomplete_candidates = sorted(set(incomplete_candidates))
    if incomplete_candidates:
        raise FileNotFoundError(
            "Incomplete CompactFaces input detected: "
            f"{[str(p) for p in incomplete_candidates]}. "
            "Expected metadata.csv, core_600png/images/, "
            "additional_optional_335png/images/, and "
            "collection_builder_187/WEB_COLLECTION_BUILDER/previews/ "
            "(or the complete documented ZIP layout). "
            "MODE=auto will not fall back to DEMO when CompactFaces data is incomplete.")
    print("KAGGLE_INPUT_CANDIDATES", [str(p) for p in candidates])
if root is None and len(candidates) == 1:
    root = candidates[0]
if root is None and len(candidates) > 1:
    raise ValueError(f"Multiple dataset roots: {candidates}. Set DATASET_DIR explicitly.")
archive_layout = "none"
if root is not None:
    if all((root / f).is_file() for f in OPTIONAL_LOCAL_ZIP_FILES):
        archive_layout = "zip"
    elif is_extracted_root(root):
        archive_layout = "extracted"
    else:
        missing = [f for f in OPTIONAL_LOCAL_ZIP_FILES if not (root / f).is_file()]
        raise FileNotFoundError(
            f"Incomplete CompactFaces input at {root}. "
            "For the Kaggle dataset, attach extracted metadata.csv, core_600png/images/, "
            "additional_optional_335png/images/, and "
            "collection_builder_187/WEB_COLLECTION_BUILDER/previews/. "
            f"For offline ZIP input, the missing optional archives are: {missing}.")
if MODE == "full" and root is None:
    diagnostics = {"input_directory": str(kaggle_inputs),
                   "input_exists": kaggle_inputs.is_dir(),
                   "top_level": [],
                   "metadata_hits": []}
    if kaggle_inputs.is_dir():
        diagnostics["top_level"] = sorted(p.name for p in kaggle_inputs.iterdir())[:30]
        diagnostics["metadata_hits"] = [
            str(p) for p in list(kaggle_inputs.rglob("metadata.csv"))[:16]]
        for directory in kaggle_inputs.iterdir():
            if directory.is_dir():
                diagnostics["top_level_" + directory.name] = sorted(
                    item.name for item in directory.iterdir())[:25]
    print("KAGGLE_INPUT_DIAGNOSTICS", json.dumps(diagnostics, ensure_ascii=False))
    raise FileNotFoundError("FULL requires attached Kaggle ZIP or extracted dataset. "
                            "See KAGGLE_INPUT_DIAGNOSTICS for actual mounted layout.")
active_mode = "demo" if MODE == "demo" or root is None else "full"
started = time.monotonic()
checks = {}
say(f"MODE: {HEADINGS[LANGUAGE][active_mode]}",
    f"CHẾ ĐỘ: {HEADINGS[LANGUAGE][active_mode]}")
if active_mode == "demo":
    say("No full dataset was selected. The sample checks are NOT a 935-image acceptance.",
        "Không chọn dataset đầy đủ. Kiểm thử mẫu KHÔNG xác minh 935 PNG gốc.")
else:
    say(f"Read-only dataset root: {root} · layout: {archive_layout}",
        f"Thư mục dữ liệu chỉ đọc: {root} · định dạng: {archive_layout}")
```

> **Tiếng Việt**


Cell này kiểm cột bắt buộc, cặp identity/slot duy nhất, collection hợp lệ, định dạng SHA, đường dẫn ảnh và mối liên hệ checksum SOURCE. FULL còn kiểm đúng số lượng Core 120 × 5 và Additional 67 × 5, checksum SHA256 không bị trùng và liên kết SOURCE hợp lệ.

Trong DEMO, mười dòng metadata được **tạo để minh họa schema**, với hash mô phỏng hợp lệ nhưng **không có ảnh PNG**. Không được xem đây là bằng chứng kiểm chứng 935 ảnh gốc.


```python
if active_mode == "full":
    with (root / "metadata.csv").open(newline="", encoding="utf-8-sig") as handle:
        reader = csv.DictReader(handle)
        require(reader.fieldnames is not None and REQUIRED_COLS <= set(reader.fieldnames),
                f"Missing metadata columns: {REQUIRED_COLS - set(reader.fieldnames or [])}")
        rows = list(reader)
else:
    rows = []
    for family, collection in (("DEMO:0001", "CORE"), ("DEMO:0002", "ADDITIONAL_OPTIONAL")):
        source_hash = hashlib.sha256(f"{family}/SOURCE/example".encode()).hexdigest()
        for slot in SLOTS:
            rows.append({"identity_key": family, "collection": collection, "slot": slot,
                         "sha256": source_hash if slot == "SOURCE" else
                                   hashlib.sha256(f"{family}/{slot}/example".encode()).hexdigest(),
                         "source_sha256": source_hash, 
                         "file_name": f"images/{family.replace(chr(58), chr(95))}_{slot}.png",
                         "archive_name": COLLECTIONS[collection][0],
                         "archive_file_name": f"images/{family.replace(':', '_')}_{slot}.png"})
families = defaultdict(dict)
collections = Counter()
seen_paths = set()
seen_imagefolder_paths = set()
for r in rows:
    require(REQUIRED_COLS <= set(r), f"Metadata row missing columns: {REQUIRED_COLS - set(r)}")
    require(safe_member(r["file_name"]), f"Invalid ImageFolder path: {r['file_name']}")
    require(r["file_name"] not in seen_imagefolder_paths,
            f"Duplicate ImageFolder path: {r['file_name']}")
    seen_imagefolder_paths.add(r["file_name"])
    identity, slot, collection = r["identity_key"], r["slot"], r["collection"]
    require(collection in COLLECTIONS, f"Invalid collection: {identity}/{collection}")
    if active_mode == "full":
        require((collection == "CORE" and re.fullmatch(r"CORE:[0-9]{4}", identity))
                or (collection == "ADDITIONAL_OPTIONAL" and re.fullmatch(r"AD:AD[0-9]{4}", identity)),
                f"Invalid identity/collection: {identity}/{collection}")
    require(slot in SLOTS and slot not in families[identity],
            f"Invalid/duplicate slot: {identity}/{slot}")
    require(is_sha(r["sha256"]) and is_sha(r["source_sha256"]),
            f"Invalid SHA256 syntax: {identity}/{slot}")
    require(safe_member(r["archive_file_name"]),
            f"Unsafe ZIP member path: {r['archive_file_name']}")
    require(r["archive_name"] == COLLECTIONS[collection][0],
            f"Wrong ZIP assignment: {identity}/{slot}")
    require((r["archive_name"], r["archive_file_name"]) not in seen_paths,
            f"Duplicate ZIP path: {identity}/{slot}")
    seen_paths.add((r["archive_name"], r["archive_file_name"]))
    families[identity][slot] = r
    collections[collection] += 1
for identity, views in families.items():
    require(set(views) == set(SLOTS), f"Missing/extra slots: {identity}")
    source_hash = views["SOURCE"]["sha256"]
    require(all(row["source_sha256"] == source_hash for row in views.values()),
            f"Inconsistent SOURCE binding: {identity}")
    require(len({row["collection"] for row in views.values()}) == 1,
            f"Mixed collections: {identity}")
if active_mode == "full":
    require(len(rows) == 935 and len(families) == 187,
            f"Wrong inventory: {len(rows)} images, {len(families)} identities")
    for collection, (_, image_count, identity_count) in COLLECTIONS.items():
        require(collections[collection] == image_count,
                f"{collection}: {collections[collection]} images instead of {image_count}")
        require(sum(views["SOURCE"]["collection"] == collection for views in families.values()) ==
                identity_count, f"Wrong {collection} identity count")
    require(len({r["sha256"] for r in rows}) == len(rows),
            "Repeated PNG SHA256 found in the active dataset")
    checks["metadata"] = "PASS"
else:
    checks["metadata"] = "DEMO_PASS"
say(f"Metadata check: {checks['metadata']} · {len(families)} families / {len(rows)} records",
    f"Kiểm metadata: {checks['metadata']} · {len(families)} bộ / {len(rows)} bản ghi")
for name in COLLECTIONS:
    say(f"  {name}: {collections[name]} records", f"  {name}: {collections[name]} bản ghi")
```

## 04 · Verify PNG SHA256 from the actual input files

> **English**

For the published **Kaggle-extracted layout**, FULL reads each file under `core_600png/images/` or `additional_optional_335png/images/`, checks its PNG signature, checks the 1024×1024 IHDR dimensions of every PNG, then streams SHA256 against `metadata.csv`. It verifies all 935 image paths without rewriting any images.

If `VERIFY_PNG_SHA256=False`, image hashes are not verified and the result is **PARTIAL**, not FULL_PASS. This notebook documents the released Kaggle folder layout.

**Runtime:** reading and hashing the distributed images can take time on Kaggle CPU, but does not use GPU quota.


<a id="integrity-vi"></a>
## 04 · Kiểm SHA256 từng PNG từ các file đầu vào thực tế

> **Tiếng Việt**

Với **cấu trúc Kaggle đã giải nén**, FULL mở từng file trong `core_600png/images/` hoặc `additional_optional_335png/images/`, kiểm chữ ký PNG, kiểm kích thước IHDR 1024×1024 của từng ảnh, rồi tính SHA256 theo `metadata.csv`. Notebook kiểm đủ 935 đường dẫn ảnh mà không ghi đè dữ liệu.

Nếu `VERIFY_PNG_SHA256=False`, notebook không kiểm byte ảnh và kết quả là **PARTIAL**, không phải FULL_PASS. Hướng dẫn này dành cho cấu trúc thư mục đang phát hành trên Kaggle.

**Hiệu năng:** đọc và hash ảnh có thể mất thời gian trên Kaggle CPU, nhưng không dùng quota GPU.


```python
verified_images = 0
verified_dimensions = 0
byte_counts = Counter()
if active_mode == "demo":
    checks["png_dimensions"] = "SKIPPED_DEMO"
    checks["png_sha256"] = "SKIPPED_DEMO"
    say("SKIPPED: no original PNG input in DEMO mode.",
        "BỎ QUA: DEMO không có input PNG gốc.")
else:
    for collection, (archive_name, count, _) in COLLECTIONS.items():
        expected_rows = [row for row in rows if row["collection"] == collection]
        expected_members = {r["archive_file_name"] for r in expected_rows}
        if archive_layout == "zip":
            with zipfile.ZipFile(root / archive_name) as archive:
                entries = archive.namelist()
                actual_members = {name for name in entries if name.lower().endswith(".png")}
                require(len(entries) == len(set(entries)), f"Duplicate ZIP paths: {archive_name}")
                require(actual_members == expected_members and len(expected_members) == count,
                        f"PNG ZIP inventory mismatch: {archive_name}")
                for n, row in enumerate(expected_rows, 1):
                    info = archive.getinfo(row["archive_file_name"])
                    byte_counts[collection] += info.file_size
                    with archive.open(info) as stream:
                        actual_hash = verify_png_header_and_hash(
                            stream, row["archive_file_name"], VERIFY_PNG_SHA256)
                    verified_dimensions += 1
                    if VERIFY_PNG_SHA256:
                        require(actual_hash == row["sha256"],
                                f"SHA256 mismatch: {row['archive_file_name']}")
                        verified_images += 1
                    if n % 125 == 0 or n == count:
                        say(f"{archive_name}: {n}/{count} inspected",
                            f"{archive_name}: đã kiểm {n}/{count}")
        elif archive_layout == "extracted":
            directory = root / Path(archive_name).stem
            actual_members = {p.relative_to(directory).as_posix()
                              for p in directory.rglob("*.png")}
            require(actual_members == expected_members and len(expected_members) == count,
                    f"Extracted PNG inventory mismatch: {directory}")
            for n, row in enumerate(expected_rows, 1):
                file = directory / row["archive_file_name"]
                byte_counts[collection] += file.stat().st_size
                with file.open("rb") as stream:
                    actual_hash = verify_png_header_and_hash(
                        stream, row["archive_file_name"], VERIFY_PNG_SHA256)
                verified_dimensions += 1
                if VERIFY_PNG_SHA256:
                    require(actual_hash == row["sha256"],
                            f"SHA256 mismatch: {file.name}")
                    verified_images += 1
                if n % 125 == 0 or n == count:
                    say(f"{directory.name}: {n}/{count} inspected",
                        f"{directory.name}: đã kiểm {n}/{count}")
        else:
            raise ValueError(f"Unknown input layout: {archive_layout}")
    require(verified_dimensions == 935, f"Incomplete PNG dimensions: {verified_dimensions}/935")
    checks["png_dimensions"] = "PASS"
    checks["png_sha256"] = "PASS" if VERIFY_PNG_SHA256 else "SKIPPED_BY_OPTION"
    say(f"PNG integrity: {checks['png_sha256']}, hashes: {verified_images}/935",
        f"Toàn vẹn PNG: {checks['png_sha256']}, SHA256 khớp: {verified_images}/935")
```

> **English**

On Kaggle, FULL checks the presence, nonempty size and exact count of **935 JPEG previews** directly under `collection_builder_187/WEB_COLLECTION_BUILDER/previews/`; Builder integrity is `PASS`.

## 05 · Kiểm JPEG Builder và phân biệt checksum của ZIP gốc

> **Tiếng Việt**

Trên Kaggle, FULL kiểm trực tiếp **935 JPEG preview** và kích thước file hợp lệ trong `collection_builder_187/WEB_COLLECTION_BUILDER/previews/`; kiểm Builder được ghi `PASS`.

Vì Kaggle Dataset Version 1 cung cấp thư mục ảnh, không cung cấp ZIP, notebook **không thể kiểm CRC và SHA256 của tệp ZIP nguyên vẹn từ input này**. Báo cáo JSON ghi `builder_crc = "SKIPPED_KAGGLE_EXTRACTED"` và `zip_files = "SKIPPED_KAGGLE_EXTRACTED"`. Đây là kết quả dự kiến, **không ngăn `FULL_PASS`**.

Các phép kiểm cấp ZIP không áp dụng cho bản phát hành hiện tại; không được nhầm với SHA256 đã kiểm của 935 PNG.

**Phạm vi Builder:** Tên của 935 JPEG phải tương ứng từng identity/slot trong metadata; từng ảnh được đọc và kiểm định dạng JPEG; catalog được đối chiếu với SHA256 và đường dẫn PNG. Kiểm tra này **không chứng minh** độ tương đồng thị giác của JPEG với PNG gốc.


```python
def expected_builder_paths(metadata_rows):
    expected = set()
    for row in metadata_rows:
        identity = row["identity_key"]
        if row["collection"] == "CORE":
            require(bool(re.fullmatch(r"CORE:[0-9]{4}", identity)),
                    f"Invalid Core identity for Builder: {identity}")
            prefix = "core_"
        else:
            require(row["collection"] == "ADDITIONAL_OPTIONAL" and
                    bool(re.fullmatch(r"AD:AD[0-9]{4}", identity)),
                    f"Invalid Additional identity for Builder: {identity}")
            prefix = "additional_"
        require(row["slot"] in SLOTS, f"Invalid Builder slot: {identity}")
        name = f"WEB_COLLECTION_BUILDER/previews/{prefix}{identity.split(':',1)[1]}_{row['slot']}.jpg"
        require(name not in expected, f"Duplicate Builder name: {name}")
        expected.add(name)
    return expected

def verify_builder_catalog(catalog, metadata_rows):
    require(catalog.get("schema") == "compactfaces_collection_builder_catalog_v1",
            "Builder catalog schema mismatch")
    groups = defaultdict(dict)
    for row in metadata_rows:
        groups[row["identity_key"]][row["slot"]] = row
    require(catalog.get("total_families") == len(groups)
            and catalog.get("total_images") == len(metadata_rows),
            "Builder catalog inventory mismatch")
    require(catalog.get("core_count") == sum(
            group["SOURCE"]["collection"] == "CORE" for group in groups.values())
            and catalog.get("additional_optional_count") == sum(
            group["SOURCE"]["collection"] == "ADDITIONAL_OPTIONAL" for group in groups.values()),
            "Builder catalog collection counts mismatch")
    entries = catalog.get("entries")
    require(isinstance(entries, list) and len(entries) == len(groups),
            "Builder catalog entries count mismatch")
    keys = [entry["key"] for entry in entries]
    require(len(keys) == len(set(keys)) and set(keys) == set(groups),
            "Builder catalog identities missing, extra or duplicated")
    for entry in entries:
        identity = entry["key"]
        group = groups[identity]
        expected_collection = "core" if group["SOURCE"]["collection"] == "CORE" else "additional"
        require(entry["collection"] == expected_collection
                and entry["id"] == identity.split(":",1)[1],
                f"Builder catalog identity mismatch: {identity}")
        require(entry["source_sha256"] == group["SOURCE"]["sha256"]
                and entry["slot_sha256"] == {slot: group[slot]["sha256"] for slot in SLOTS}
                and entry["image_paths"] == {slot: group[slot]["archive_file_name"] for slot in SLOTS},
                f"Builder catalog source/hash/path mismatch: {identity}")

def verify_builder_jpeg(handle, name):
    from PIL import Image
    with Image.open(handle) as im:
        require(im.format == "JPEG" and im.width > 0 and im.height > 0,
                f"Invalid Builder JPEG: {name}")
        im.load()  # Force full JPEG pixel decoding; verify() alone checks headers.

preview_count = 0
if active_mode == "demo":
    checks["builder"] = "SKIPPED_DEMO"
    checks["builder_crc"] = "SKIPPED_DEMO"
    checks["zip_files"] = "SKIPPED_DEMO"
    say("SKIPPED Builder: no original input.", "BỎ QUA Builder: DEMO không có ảnh gốc.")
else:
    if VERIFY_BUILDER:
        expected = expected_builder_paths(rows)
        require(len(expected) == 935, "Expected 935 unique Builder names")
        if archive_layout == "zip":
            with zipfile.ZipFile(root / "collection_builder_187.zip") as archive:
                require(archive.testzip() is None, "Builder ZIP CRC failure")
                names = archive.namelist()
                require(len(names) == len(set(names)), "Duplicate Builder ZIP paths")
                actual = {n for n in names if n.lower().endswith((".jpg",".jpeg"))}
                require(actual == expected, "Builder JPEG filenames differ from metadata identity/slot pairs")
                verify_builder_catalog(json.loads(archive.read("COLLECTION_BUILDER_CATALOG.json")), rows)
                for name in sorted(expected):
                    with archive.open(name) as jpeg:
                        verify_builder_jpeg(jpeg, name)
                    preview_count += 1
            checks["builder_crc"] = "PASS"
        else:
            base = root / "collection_builder_187"
            jpgs = [p for p in base.rglob("*") if p.is_file() and p.suffix.lower() in (".jpg",".jpeg")]
            actual = {p.relative_to(base).as_posix() for p in jpgs}
            require(actual == expected, "Extracted Builder JPEG filenames differ from metadata")
            verify_builder_catalog(json.loads((base/"COLLECTION_BUILDER_CATALOG.json").read_text(encoding="utf-8")), rows)
            for name in sorted(expected):
                with (base/name).open("rb") as jpeg:
                    verify_builder_jpeg(jpeg, name)
                preview_count += 1
            checks["builder_crc"] = "SKIPPED_KAGGLE_EXTRACTED"
        require(preview_count == 935, "Builder JPEG decoding incomplete")
        checks["builder"] = "PASS"
    else:
        checks["builder"] = "SKIPPED_BY_OPTION"
        checks["builder_crc"] = "SKIPPED_BY_OPTION"
    if archive_layout == "extracted":
        checks["zip_files"] = "SKIPPED_KAGGLE_EXTRACTED"
        say("Whole-ZIP SHA256 is unavailable for extracted Kaggle inputs.",
            "Không thể kiểm SHA256 của ZIP gốc trên các thư mục Kaggle đã giải nén.")
    elif VERIFY_DISTRIBUTION_ZIPS:
        manifest = root / "SHA256SUMS.txt"
        require(manifest.is_file(), "SHA256SUMS.txt absent")
        recorded = {}
        for line in manifest.read_text().splitlines():
            if not line.strip():
                continue
            match = re.fullmatch(r"([0-9a-f]{64})\s+\*?([A-Za-z0-9][A-Za-z0-9_.-]*)", line.strip())
            require(match is not None, f"Invalid SHA256 manifest entry: {line}")
            filename = match.group(2)
            require(filename not in recorded, "Duplicate SHA256 manifest entry")
            recorded[filename] = match.group(1)
        for filename in ("core_600png.zip","additional_optional_335png.zip","collection_builder_187.zip"):
            require(filename in recorded, f"Missing ZIP SHA256: {filename}")
        for filename,digest in recorded.items():
            with (root/filename).open("rb") as stream:
                require(sha256_stream(stream)==digest, f"ZIP SHA256 mismatch: {filename}")
        checks["zip_files"]="PASS"
    else:
        checks["zip_files"]="SKIPPED_BY_OPTION"
    say(f"Builder: {checks['builder']} · {preview_count} verified JPEGs · ZIP CRC: {checks['builder_crc']}",
        f"Builder: {checks['builder']} · {preview_count} JPEG đã kiểm · CRC ZIP: {checks['builder_crc']}")
```

## 07 · Validate prompts, seeds and SOURCE provenance

> **English**  
Cross-check 935 machine-readable prompt/seed records against PNG embedded SDCpp metadata; validate 187 SOURCE lineage records without making ancestor claims. Read-only; required for FULL_PASS.

## 07 · Kiểm chứng prompt, seed và nguồn gốc SOURCE

> **Tiếng Việt**  
Đối chiếu 935 bản ghi prompt/seed với SDCpp metadata nhúng trong PNG; kiểm 187 bản ghi SOURCE lineage, không suy đoán ảnh tổ tiên. Chỉ đọc; bắt buộc để FULL_PASS.

```python

def verify_lineage_identity_keys(lineage_entries, source_keys):
    keys = [entry["identity_key"] for entry in lineage_entries]
    require(len(keys) == len(set(keys)) and set(keys) == set(source_keys),
            "SOURCE lineage identity keys missing, extra or duplicated")

provenance_verified = 0
source_lineage_verified = 0
if active_mode == "demo":
    checks["provenance"] = "SKIPPED_DEMO"
    say("SKIPPED: provenance requires the full dataset.",
        "BỎ QUA: provenance cần dataset đầy đủ.")
else:
    from PIL import Image
    for name in ("generation_metadata.jsonl","source_lineage.jsonl",
                 "generation_metadata_audit.json"):
        require((root / name).is_file(), f"Missing provenance file: {name}")
    generated = [json.loads(line) for line in
                 (root / "generation_metadata.jsonl").read_text(encoding="utf-8").splitlines()
                 if line.strip()]
    lineage = [json.loads(line) for line in
               (root / "source_lineage.jsonl").read_text(encoding="utf-8").splitlines()
               if line.strip()]
    audit = json.loads((root / "generation_metadata_audit.json").read_text(encoding="utf-8"))
    require(len(generated) == 935 and len(lineage) == 187, "Wrong provenance record count")
    require(audit["status"] == "PASS", "Generation provenance audit is not PASS")
    for name in ("generation_metadata.jsonl","source_lineage.jsonl"):
        actual = hashlib.sha256((root / name).read_bytes()).hexdigest()
        require(actual == audit["artifacts_sha256"][name], f"JSONL digest mismatch: {name}")
    index = {(p["identity_key"],p["slot"]):p for p in generated}
    require(len(index) == 935, "Duplicate identity/slot in generation metadata")
    for entry in rows:
        key = (entry["identity_key"],entry["slot"])
        p = index[key]
        require(p["png_sha256"] == entry["sha256"]
                and p["file_name"] == entry["file_name"],
                f"Provenance CSV crosswalk mismatch: {key}")
        if archive_layout == "zip":
            with zipfile.ZipFile(root / entry["archive_name"]) as sourcezip:
                with sourcezip.open(entry["archive_file_name"]) as handle:
                    with Image.open(handle) as img:
                        embedded = img.info["parameters"]
        else:
            path = root / Path(entry["archive_name"]).stem / entry["archive_file_name"]
            with Image.open(path) as img:
                embedded = img.info["parameters"]
        require(hashlib.sha256(embedded.encode("utf-8")).hexdigest()
                == p["embedded_parameters_sha256"], f"Embedded SHA mismatch: {key}")
        require("SDCPP: " in embedded, f"SDCPP parameters missing: {key}")
        raw = json.loads(embedded.split("SDCPP: ",1)[1])
        require(raw == p["sdcpp_parameters"], f"SDCPP metadata mismatch: {key}")
        require(raw["prompt"]["positive"] == p["positive_prompt"]
                and raw["prompt"]["negative"] == p["negative_prompt"]
                and raw["seed"] == p["seed"]
                and raw["sampling"]["steps"] == p["steps"],
                f"Prompt/seed/settings mismatch: {key}")
        provenance_verified += 1
    sources = {p["identity_key"]:p for p in generated if p["slot"] == "SOURCE"}
    require(len(sources) == 187, "SOURCE provenance family count mismatch")
    verify_lineage_identity_keys(lineage, sources)
    for entry in lineage:
        s=sources[entry["identity_key"]]
        require(entry["source_png_sha256"] == s["png_sha256"]
                and entry["final_step_prompt_sha256"] == s["prompt_sha256"]
                and entry["final_step_seed"] == s["seed"],
                "SOURCE lineage final-step mismatch")
        source_lineage_verified += 1
    require(provenance_verified == 935 and source_lineage_verified == 187,
            "Provenance coverage incomplete")
    checks["provenance"] = "PASS"
    say("PASS: 935 embedded PNG provenance records and 187 SOURCE lineages.",
        "ĐẠT: 935 bản ghi metadata PNG nhúng và 187 bản ghi SOURCE lineage.")
```
